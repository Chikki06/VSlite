```python
import os
import tifffile as tiff
import numpy as np
from skimage.transform import resize

test = []
files = [f for f in os.listdir('imagesT/') if f.endswith('.tif')]
test_file_path = "/home/kia/myspinner/Mayo/Datasets/testset_Dec_17_2024.txt"
with open(test_file_path, 'r') as file:
    TEST = [line.strip() for line in file.readlines()]
if not os.path.exists('imagesT_low'):
    os.makedirs('imagesT_low')

for file in files:
    if not os.path.exists(f'imagesT_low/{file}'):
        img = tiff.imread(f'imagesT/{file}')

        img = img[::10,::10,:]



        tiff.imwrite(f'imagesT_low/{file}', img)
    he_file = f"{file[:-4]}"
    if he_file in TEST:
    
        test.append(he_file)
test.sort()
```
